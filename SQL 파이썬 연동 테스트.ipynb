```python
import pymysql

# pip install pymysql

# DB 에 저장하기
mysql = pymysql.connect(host="127.0.0.1", port = 9090 , 
                            # mysql -u root -p (mysql)
                            # create database ive_testdb;
        user="root",         # CREATE USER 'ive'@'localhost' IDENTIFIED BY 'mysql';
        # GRANT ALL PRIVILEGES ON ive_testdb.* to 'ive'@'localhost';
        # FLUSH PRIVILEGES;
        # SHOW GRANTS FOR 'ive'@'localhost';
        passwd="root", 
        db="myschool", 
        charset="utf8mb4",
        use_unicode=True    # 유니코드 사용 강제
)

cursor = mysql.cursor() 

# ★ 연결 직후 이 명령을 실행해서 세션 설정을 강제로 바꿉니다.
cursor.execute("SET NAMES utf8mb4")
cursor.execute("SET CHARACTER SET utf8mb4")
cursor.execute("SET character_set_connection = utf8mb4")
# utf8mb4 : 특수문자 포함

쿼리 = r''' 
show tables;
'''
cursor.execute(쿼리)

tables = cursor.fetchall()

# 결과 출력
for table in tables:
    print(table[0])  # 테이블 이름만 깔끔하게 출력
```

```python
쿼리 = r'''
desc students;
'''

cursor.execute(쿼리)

result = cursor.fetchall()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```
