```python
import pymysql

# pip install pymysql

# DB 에 저장하기
mysql = pymysql.connect(host="127.0.0.1", port = 9090 , 
                            # mysql -u root -p (mysql)
                            # create database myschool;
        user="root",         # CREATE USER 'ive'@'localhost' IDENTIFIED BY 'mysql';
        # GRANT ALL PRIVILEGES ON myschool.* to 'ive'@'localhost';
        # FLUSH PRIVILEGES;
        # SHOW GRANTS FOR 'ive'@'localhost';
        passwd="root", 
        db="myschool", 
        charset="utf8mb4",
        use_unicode=True    # 유니코드 사용 강제
)

cursor = mysql.cursor() 

# ★ 연결 직후 이 명령을 실행해서 세션 설정을 강제로 바꿉니다.
cursor.execute("SET NAMES utf8mb4")
cursor.execute("SET CHARACTER SET utf8mb4")
cursor.execute("SET character_set_connection = utf8mb4")
# utf8mb4 : 특수문자 포함

# 테이블 이름 출력
쿼리 = r''' 
show tables;
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력
for row in result:
    print(row[0])
```

```python
# professors 테이블에서 name , position 칼럼 (이 때 position 은 '직업' 별칭)
# students 테이블에서 name , status 칼럼 (이 때 status 에는 '생' 글자가 추가됨)
# 두 개를 행 기준으로 아래로 나란히 결합
# 참고로 concat 은 글자 결합

쿼리 = r'''
select name , position as '직업' from professors union select name, concat (status, "생")  from students;
-- 글자 결합 포함
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:5]:
    print(row)  # 튜플로 출력됩니다.

print()

# 이거는 추가로 가장 뒤에 꺼 기준으로 출력
for row in result[-6:-1]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# left 및 right 통해서 글자 분리
# 정확히는 왼쪽에서 (또는 오른쪽에서) 부터 몇 글자 까지
# 성 / 이름 별칭 사용

쿼리 = r"""
select name , left(name , 1) as 성 , right(name , 2) as 이름 , position from professors where position = '부교수';
"""

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# LPAD 통해서 사이에 뭔가 붙이고 RPAD 통해서 끝에 공백 만들기

쿼리 = r"""
select concat('[' , LPAD(id , 5 , '0') , ']' , RPAD(name , 10 , ' ')) as label from professors where department_id = 101 order by label;
"""

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# phone 가운데 번호 추출
# substring_index 2 번 사용하여 '-' 기준으로 순서 지정한다
# -1 사용해서 뒤를 이어가는 게 사실상 큰 핵심임
# substring_index(phone , '-' , -1) 은 오른쪽 한 부분을 의미 (예를 들어 010-1234-5678 에서 5678)

쿼리 = r"""
select phone , substring_index(phone , '-' , 2) as 1차시도 , substring_index(substring_index(phone , '-' , 2) , '-' , -1) as 센터 from students;
-- substring_index(phone , '-' , -1) 은 오른쪽 한 부분을 의미 (예를 들어 010-1234-5678 에서 5678)
"""

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# power : 제곱
# round : 소수점 반올림

쿼리 = r"""
select name , weight , height , round(weight / power (height / 100 , 2) , 2) as bmi from students where grade = 1 order by bmi;
"""

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# ceil : 올림
# floor : 내림
# truncate : 버림

쿼리 = r"""
 select name , weight , height , round(weight / power (height / 100 , 2) , 4) as bmi , 
 ceil(weight / power(height / 100 , 2)) as bmi_up , 
 floor(weight / power(height / 100 , 2)) as bmi_down , 
 truncate(weight / power(height / 100 , 2) , 3) 
 as bmi_trun from students where grade = 2 order by name;
"""

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:10]:
    print(row)  # 튜플로 출력됩니다.
```
