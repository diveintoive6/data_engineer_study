```python
import pymysql

# pip install pymysql

# DB 에 저장하기
mysql = pymysql.connect(host="127.0.0.1", port = 9090 , 
                            # mysql -u root -p (mysql)
                            # create database myschool;
        user="root",         # CREATE USER 'ive'@'localhost' IDENTIFIED BY 'mysql';
        # GRANT ALL PRIVILEGES ON myschool.* to 'ive'@'localhost';
        # FLUSH PRIVILEGES;
        # SHOW GRANTS FOR 'ive'@'localhost';
        passwd="root", 
        db="myschool", 
        charset="utf8mb4",
        use_unicode=True    # 유니코드 사용 강제
)

cursor = mysql.cursor() 

# ★ 연결 직후 이 명령을 실행해서 세션 설정을 강제로 바꿉니다.
cursor.execute("SET NAMES utf8mb4")
cursor.execute("SET CHARACTER SET utf8mb4")
cursor.execute("SET character_set_connection = utf8mb4")
# utf8mb4 : 특수문자 포함

# 테이블 이름 출력
쿼리 = r''' 
show tables;
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력
for row in result:
    print(row[0])
```

```python
# 만 나이 공식 사용
# 단 이 방식은 만 나이 기준이며 실제 나이 기준은 + 1 을 해야 하기 때문에 년을 먼저 기준으로 잡는다
# now() 통해서 현재 시점의 날짜와 시간 파악 가능

쿼리 = r'''
select name , birthdate , floor(datediff(now() , birthdate) / 365) as '만 나이' from students order by birthdate limit 15;

-- 이 방식은 만 나이 기준이며 실제 나이 기준은 + 1 을 해야 하기 때문에 년을 먼저 기준으로 잡는다
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:15]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# 실제 나이 공식 사용
# 단 이 방식은 만 나이 기준이며 실제 나이 기준은 + 1 을 해야 하기 때문에 년을 먼저 기준으로 잡는다
# now() 통해서 현재 시점의 날짜와 시간 파악 가능

쿼리 = r'''
select name , birthdate , year(now()) - year(birthdate) + 1 as '실제 나이' from students order by birthdate limit 20;

-- 이 방식은 만 나이 기준이며 실제 나이 기준은 + 1 을 해야 하기 때문에 년을 먼저 기준으로 잡는다
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:15]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# GROUPING 그룹핑
# group by 사용

쿼리 = r'''
select department_id , count(*) , count(comm) from professors group by department_id order by department_id;

-- count(*) 는 NULL 까지 포함해서 전체를 세지만 count() 에 컬럼이 지정된 경우 NULL 은 절대로 세지 않는다
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:15]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# GROUPING 그룹핑
# group by 사용

쿼리 = r'''
select department_id , grade , count(*) , avg(height) , avg(weight) from students 
group by department_id , grade order by department_id , grade;

-- 그룹에 대한 값은 집계된 값만 표시가 가능하다
-- 예를 들어 아이브 그룹을 만든다고 했을 때 이서 , 레이 같이 멤버명을 모두 표기할 수는 없지만 평균 키를 표기 할 수 있다
-- 멤버명을 모두 표기한다 = 1 대 1 매칭 위반!!
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:15]:
    print(row)  # 튜플로 출력됩니다.
```
