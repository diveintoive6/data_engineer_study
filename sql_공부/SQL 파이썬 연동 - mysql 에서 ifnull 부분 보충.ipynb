```python
import pymysql

# pip install pymysql

# DB 에 저장하기
mysql = pymysql.connect(host="127.0.0.1", port = 9090 , 
                            # mysql -u root -p (mysql)
                            # create database myschool;
        user="root",         # CREATE USER 'ive'@'localhost' IDENTIFIED BY 'mysql';
        # GRANT ALL PRIVILEGES ON myschool.* to 'ive'@'localhost';
        # FLUSH PRIVILEGES;
        # SHOW GRANTS FOR 'ive'@'localhost';
        passwd="root", 
        db="myschool", 
        charset="utf8mb4",
        use_unicode=True    # 유니코드 사용 강제
)

cursor = mysql.cursor() 

# ★ 연결 직후 이 명령을 실행해서 세션 설정을 강제로 바꿉니다.
cursor.execute("SET NAMES utf8mb4")
cursor.execute("SET CHARACTER SET utf8mb4")
cursor.execute("SET character_set_connection = utf8mb4")
# utf8mb4 : 특수문자 포함

# 테이블 이름 출력
쿼리 = r''' 
show tables;
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력
for row in result:
    print(row[0])
```

```python
# ifnull 통해서 Null 이면 값 대체하기
# is null 을 통하여 확실하게 확인이 가능함

쿼리 = r'''
select student_id , subject_id , score , ifnull(score , '미입력') as '성적' from enrollments where score is null;

-- 점수가 없다면 미입력으로 표기
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# ifnull 통해서 Null 이면 값 대체하기
# ifnull 의 실제 원리

쿼리 = r'''
select student_id , subject_id , score , if(score is null , '미입력' , score) as '성적' from enrollments;

-- ifnull 의 원리
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# ifnull 통해서 Null 이면 값 대체하기
# ifnull 의 실제 원리

쿼리 = r'''
select student_id , subject_id , score , if(score is null , '미입력' , score) as '성적' from enrollments where score is null;

-- ifnull 의 원리
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# 따라서 IFNULL (가장 흔하게 사용) 은 지정한 표현식이 NULL인지 확인하고, 
# NULL이 아니면 그 값을, NULL이면 지정한 대체값을 (2번째 인자를) 반환

# COALESCE (표준 SQL 함수)
# 나열된 인수들 중에서 NULL이 아닌 첫 번째 값을 반환합니다. 2개 이상의 인수를 받을 수 있어 매우 유연
```

```python
# 여기서는 score 가 NULL 이어도 subject_id 가 존재하기 때문에 아무 정보 없음은 사실상 안 나온다
# 즉 , COALESCE 는 NULL이 아닌 첫 번째 값을 반환

쿼리 = r'''
SELECT COALESCE(score , subject_id , '아무 정보 없음') FROM enrollments;
'''

cursor.execute(쿼리)

# cursor.description의 각 튜플에서 첫 번째 값 (index 0) 이 칼럼명
column_names = [col[0] for col in cursor.description]

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# Oracle의 NVL2(expr1, val2, val3)는 expr1이 NULL이 아니면 val2를, 
# NULL이면 val3를 반환하는 함수입니다. 
# MySQL에서는 IF 함수나 CASE 문으로 대체 가능

쿼리 = r'''
SELECT IF(score IS NOT NULL, '점수 ok', '시험 안 봄') FROM enrollments;
'''

cursor.execute(쿼리)

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```

```python
# Oracle의 NVL2(expr1, val2, val3)는 expr1이 NULL이 아니면 val2를, 
# NULL이면 val3를 반환하는 함수입니다. 
# MySQL에서는 IF 함수나 CASE 문으로 대체 가능

쿼리 = r'''
SELECT CASE 
    WHEN score IS NOT NULL THEN '점수 ok' 
    ELSE '시험 안 봄'
END FROM enrollments;
'''

cursor.execute(쿼리)

result = cursor.fetchall()

# 칼럼명 출력하기
print(column_names)
print()

# 결과 출력하기
for row in result[:20]:
    print(row)  # 튜플로 출력됩니다.
```
